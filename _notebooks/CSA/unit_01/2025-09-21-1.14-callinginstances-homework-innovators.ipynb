## Popcorn Hack #2

- Add on to the previous counter class and add three more methods for subtraction, multiplication, and division.

- Call all five methods outside of the class.

```java
// CODE_RUNNER: Popcorn Hack #2: add subtract, multiply and divide, then call all five methods.

// Work by Dhyan Soni

class Counter {
    private int count;

    public void add(int x) {
        count += x;
    }

    public void subtract(int x) {
        count -= x;
    }

    public void multiply(int x) {
        count *= x;
    }

    public void divide(int x) {
        if (x == 0) {
            System.out.println("Cannot divide by zero, skipping.");
            return;
        }
        count /= x;
    }

    public int getCount() {
        return count;
    }
}

public class CounterTest {
    public static void main(String[] args) {
        Counter c = new Counter();

        c.add(20);
        System.out.println("after add(20):      " + c.getCount());

        c.subtract(6);
        System.out.println("after subtract(6):  " + c.getCount());

        c.multiply(3);
        System.out.println("after multiply(3):  " + c.getCount());

        c.divide(2);
        System.out.println("after divide(2):    " + c.getCount());

        c.divide(0);
        System.out.println("after divide(0):    " + c.getCount());
    }
}

CounterTest.main(null);
```

`scale(2)` doubles both fields, so length becomes 6 and width becomes 8, and `getArea()` returns

the class name, and E has no object at all. Only D goes through the public accessor on a real

## My Homework Hack: Student Grade Tracker

```java
// CODE_RUNNER: Homework Hack: Student Grade Tracker. Add a third student or more grades.

// Homework Hack: Student Grade Tracker

public class Student {
    private String name;
    private int totalPoints;
    private int numAssignments;

    public Student(String name) {
        this.name = name;
        this.totalPoints = 0;
        this.numAssignments = 0;
    }

    // void: records one grade and updates both running totals
    public void addGrade(int points) {
        totalPoints += points;
        numAssignments++;
        System.out.println("Adding grade: " + points + " points");
    }

    // non-void: cast first so the division is not integer division
    public double getAverage() {
        if (numAssignments == 0) {
            return 0.0;
        }
        return Math.round((double)totalPoints / numAssignments * 100.0) / 100.0;
    }

    // non-void: turns the average into a letter
    public String getLetterGrade() {
        double average = getAverage();
        if (average >= 90) return "A";
        if (average >= 80) return "B";
        if (average >= 70) return "C";
        if (average >= 60) return "D";
        return "F";
    }

    // void: prints the formatted report
    public void printReport() {
        System.out.println("\n--- " + name.split(" ")[0] + "'s Grade Report ---");
        System.out.println("Student Name: " + name);
        System.out.println("Total Points: " + totalPoints);
        System.out.println("Assignments Completed: " + numAssignments);
        System.out.println("Current Average: " + getAverage());
        System.out.println("Letter Grade: " + getLetterGrade());

        String letter = getLetterGrade();
        if (letter.equals("A") || letter.equals("B")) {
            System.out.println("Status: Excellent work!");
        } else if (letter.equals("C")) {
            System.out.println("Status: Keep working hard!");
        } else {
            System.out.println("Status: Let's set up a check-in.");
        }
    }

    public String getName() {
        return name;
    }

    public static void main(String[] args) {
        System.out.println("=== Student Grade Tracker System ===");

        System.out.println("\nCreating student: Emma Rodriguez");
        Student emma = new Student("Emma Rodriguez");
        System.out.println("Student created successfully!");

        System.out.println("\n--- Adding Grades for Emma ---");
        emma.addGrade(95);
        emma.addGrade(88);
        emma.addGrade(92);
        emma.addGrade(85);
        emma.printReport();

        System.out.println("\n========================================");

        System.out.println("\nCreating student: James Wilson");
        Student james = new Student("James Wilson");
        System.out.println("Student created successfully!");

        System.out.println("\n--- Adding Grades for James ---");
        james.addGrade(78);
        james.addGrade(82);
        james.addGrade(75);
        james.printReport();

        System.out.println("\n========================================");

        System.out.println("\nFinal Summary:");
        System.out.println(emma.getName() + " - Average: " + emma.getAverage()
                + " (" + emma.getLetterGrade() + ")");
        System.out.println(james.getName() + " - Average: " + james.getAverage()
                + " (" + james.getLetterGrade() + ")");
    }
}

Student.main(null);
```

### Why each method is where it is

Every one of these is an **instance** method, because every one of them reads or writes a single
student's own `totalPoints` and `numAssignments`. Emma and James each get their own copy of those
fields, which is the whole point of the assignment: calling `emma.addGrade(95)` cannot possibly
disturb James's totals.

Two details worth pointing out:

- `getAverage()` casts `totalPoints` to `double` *before* dividing. Without the cast, James's
  `235 / 3` would be integer division and report 78 instead of 78.33, which would not change his
  letter grade here but absolutely could on a borderline case.
- `printReport()` calls `getAverage()` and `getLetterGrade()` rather than recomputing them. If the
  grading scale ever changes, there is exactly one place to edit.
