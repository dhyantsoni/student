## Unit 1: Primitive Types

### Question 1

Which of the following is a valid declaration of a variable of type int in Java?<br>
a) int 123variable;<br>
b) int variable123;<br>
c) int variable#123;<br>
d) int variable 123;

Answer: b) int variable123;


```java
// CODE_RUNNER: Q1 Hack: Java naming conventions. Uncomment an illegal name to see the compiler complain.

// Java uses camelCase for variables and methods, PascalCase for classes,
// and SCREAMING_SNAKE_CASE for final constants. Not snake_case.

public class NamingConventions {
    public static void main(String[] args) {
        int variable123 = 123;              // valid: starts with a letter
        int studentGradeLevel = 11;         // camelCase, the convention for variables
        final int MAX_STUDENTS = 30;        // constants are ALL_CAPS with underscores
        String firstName = "Dhyan";
        double gpaThisSemester = 3.95;
        boolean isEnrolled = true;
        int _underscoreStart = 1;           // legal, but not conventional
        int $dollarStart = 2;               // also legal, also not conventional

        System.out.println(variable123);
        System.out.println(studentGradeLevel + " " + MAX_STUDENTS + " " + firstName);
        System.out.println(gpaThisSemester + " " + isEnrolled);
        System.out.println(_underscoreStart + " " + $dollarStart);

        // These would NOT compile:
        // int 123variable;    -- cannot start with a digit
        // int variable#123;   -- # is not a legal identifier character
        // int variable 123;   -- identifiers cannot contain spaces
        // int class;          -- 'class' is a reserved keyword
    }
}

NamingConventions.main(null);
```

### Question 2

What is the value of the following expression in Java: 5 / 2?<br>
a) 2.5<br>
b) 3<br>
c) 2<br>
d) 2.0

Answer: c) 2


```java
// CODE_RUNNER: Q2 Hacks: integer vs floating point division, and how each number type behaves.

public class NumberTypes {
    public static void main(String[] args) {
        // Q2.1: integer vs floating point division
        System.out.println("5 / 2         = " + (5 / 2));            // 2, both operands are int
        System.out.println("5.0 / 2       = " + (5.0 / 2));          // 2.5
        System.out.println("5 / 2.0       = " + (5 / 2.0));          // 2.5
        System.out.println("(double)5 / 2 = " + ((double) 5 / 2));   // 2.5, cast before dividing
        System.out.println("(double)(5/2) = " + ((double)(5 / 2))); // 2.0, too late
        System.out.println("5 % 2         = " + (5 % 2));            // 1, the dropped remainder

        // Q2.2: the number types, declaration and assignment
        byte  smallCount = 100;              // 8 bits,  -128 to 127
        short mediumCount = 30000;           // 16 bits
        int   population = 2147483647;       // 32 bits, the default whole number type
        long  worldPopulation = 8100000000L; // 64 bits, needs the L suffix
        float ratio = 3.14f;                 // 32 bits, needs the f suffix
        double precise = 3.141592653589793;  // 64 bits, the default decimal type

        System.out.println("\nbyte:   " + smallCount);
        System.out.println("short:  " + mediumCount);
        System.out.println("int:    " + population);
        System.out.println("long:   " + worldPopulation);
        System.out.println("float:  " + ratio);
        System.out.println("double: " + precise);

        double widened = population;      // widening is automatic
        int narrowed = (int) precise;     // narrowing needs an explicit cast
        System.out.println("\nint widened to double:  " + widened);
        System.out.println("double narrowed to int: " + narrowed);

        System.out.println("Integer.MAX_VALUE + 1 = " + (population + 1));  // wraps around
    }
}

NumberTypes.main(null);
```

### Question 3

Which primitive type is used to represent a single character in Java?<br>
a) char<br>
b) String<br>
c) int<br>
d) byte

Answer: a) char

```java
// CODE_RUNNER: Q3 Hacks: the non-number primitives, and how String behaves as a reference type.

public class NonNumberTypes {
    public static void main(String[] args) {
        // Q3.1: the non-number primitives
        boolean isEnrolled = true;      // only true or false
        boolean isGraduated = false;
        char initial = 'D';             // exactly ONE character, single quotes
        char digitChar = '7';           // a character, not the number 7

        System.out.println("boolean isEnrolled:  " + isEnrolled);
        System.out.println("boolean flipped:     " + !isEnrolled);
        System.out.println("boolean AND:         " + (isEnrolled && isGraduated));
        System.out.println("char initial:        " + initial);

        // a char is secretly a number: it holds a Unicode code point
        System.out.println("initial as an int:   " + (int) initial);        // 68
        System.out.println("initial + 1 as char: " + (char)(initial + 1));  // E
        System.out.println("digitChar '7' + 1:   " + (digitChar + 1));      // 56, NOT 8

        // Q3.2: the String type
        String name = "Dhyan";               // String is a CLASS, not a primitive
        String same = "Dhyan";               // both literals share one pooled object
        String built = new String("Dhyan");  // 'new' forces a separate object

        System.out.println("\nname.length():       " + name.length());
        System.out.println("name.toUpperCase():  " + name.toUpperCase());
        System.out.println("name.substring(0,2): " + name.substring(0, 2));
        System.out.println("name.charAt(0):      " + name.charAt(0));

        System.out.println("\nname == same:       " + (name == same));      // true, pooled
        System.out.println("name == built:      " + (name == built));     // false
        System.out.println("name.equals(built): " + name.equals(built));  // true

        // Strings are immutable: methods return a NEW string
        name.toUpperCase();
        System.out.println("\nname after an unassigned toUpperCase(): " + name);

        String nothing = null;   // unlike a primitive, a reference can be null
        System.out.println("a String can be null: " + nothing);
    }
}

NonNumberTypes.main(null);
```

### Question 4

Answer the following questions based on the code cell:
* a) What kind of types are person1 and person2? 
* Answer: **Reference types.** They are `Person` variables, so each one stores the memory address of a `Person` object rather than the object itself. Everything that is not one of the eight primitives is a reference type.
* b) Do person1 and person3 point to the same value in memory?
* Answer: **Yes.** `Person person3 = person1;` copies the address, not the object. No `new` keyword appears on that line, so no second `Person` is created and both variables point at the same object. Changing `person3.name` would also change `person1.name`, and `person1 == person3` is `true`.
* c) Is the integer "number" stored in the heap or in the stack?
* Answer: **The stack.** `number` is a local primitive declared inside `main`, so it lives in `main`'s stack frame and holds the value 16 directly. It disappears when `main` returns.
* d) Is the value that "person1" points to stored in the heap or in the stack?
* Answer: **The heap.** The *reference* `person1` is on the stack, but the `Person` object it points to was built with `new`, and everything built with `new` goes on the heap. That split is the reason part (b) works the way it does.

```java
public class Person {
    String name;
    int age;
    int height;
    String job;

    public Person(String name, int age, int height, String job) {
        this.name = name;
        this.age = age;
        this.height = height;
        this.job = job;
    }
}

public static void main(String[] args) {
    Person person1 = new Person("Carl", 25, 165, "Construction Worker");
    Person person2 = new Person("Adam", 29, 160, "Truck Driver");
    Person person3 = person1;
    int number = 16;
    System.out.println(number);
}
main(null); // This is required in Jupiter Notebook to run the main method.
```

### Q5 (a) and (b): primitive types vs reference types

**Primitive types** (`int`, `double`, `boolean`, `char`, `long`, `short`, `byte`, `float`) hold the
actual value. In this banking application, `balance` and `accountNumber` are primitives, because a
balance *is* a number, and there is no behavior to attach to it.

**Reference types** (`String`, `Account`, `ArrayList`, and every class) hold the memory address of
an object. `name` is a reference type, and so is every `Account` object itself.

**Where things live in memory:**

| Location | What is stored | Example from the code |
|---|---|---|
| **Stack** | Local variables and parameters, including references to objects | `carl`, `adam`, `nina`, and `carlAlias` inside `main` each hold an address |
| **Heap** | The objects themselves, including the primitive instance variables inside them | the three `Account` objects, each carrying its own `balance` and `accountNumber`, plus the `String` objects for the names |
| **Static area** | One shared copy per class, not per instance | `lastAccountNumber` and `interestRate` |

**Instances.** Each `new Account(...)` creates a separate instance on the heap with its own
`balance`, which is why Carl earning interest does not touch Adam's account. `lastAccountNumber` is
the opposite: it is `static`, so all three constructors increment the *same* counter, which is
exactly what makes the account numbers unique.

**The alias at the end of `main`** is the part worth staring at. `carlAlias = carl` copies the
address, not the account. There is still only one Carl account on the heap, so zeroing the balance
through `carlAlias` also zeroes it through `carl`. If `balance` were a primitive being copied
instead, that would not have happened. This is the single most tested idea in Unit 1.

```java
// CODE_RUNNER: Q5 Hack: the banking application. Add a fourth account and watch the numbers increment.

// Q5 Hack: banking application demonstrating primitive vs reference types

public class Account {
    // Reference type. 'name' lives on the stack as a reference; the String
    // object it points to lives on the heap.
    private String name;

    // Primitives. Because they are instance variables they sit inside the
    // Account object on the heap, and each Account gets its own copy.
    private double balance;
    private int accountNumber;

    // Static: one copy for the whole class, not per instance.
    private static int lastAccountNumber = 1000;
    private static double interestRate = 0.05;

    // two argument constructor
    public Account(String name, double balance) {
        this.name = name;
        this.balance = balance;

        lastAccountNumber++;                  // shared counter guarantees uniqueness
        this.accountNumber = lastAccountNumber;
    }

    public double getBalance() {
        return balance;
    }

    public void setBalance(double balance) {
        this.balance = balance;
    }

    public double calculateInterest() {
        double interest = balance * interestRate;
        setBalance(getBalance() + interest);
        return interest;
    }

    public static void setInterestRate(double rate) {
        interestRate = rate;
    }

    public void printAccount() {
        System.out.printf("#%d  %-8s balance: $%.2f%n", accountNumber, name, balance);
    }

    public static void main(String[] args) {
        Account carl = new Account("Carl", 2500.00);
        Account adam = new Account("Adam", 1200.00);
        Account nina = new Account("Nina", 800.00);

        System.out.println("Accounts created (note the auto-incrementing numbers):");
        carl.printAccount();
        adam.printAccount();
        nina.printAccount();

        System.out.println("\nApplying interest at " + (interestRate * 100) + "%:");
        System.out.printf("Carl earned $%.2f%n", carl.calculateInterest());
        System.out.printf("Adam earned $%.2f%n", adam.calculateInterest());
        System.out.printf("Nina earned $%.2f%n", nina.calculateInterest());

        System.out.println("\nAfter interest:");
        carl.printAccount();
        adam.printAccount();
        nina.printAccount();

        // Reference behavior: an alias is not a copy
        Account carlAlias = carl;
        carlAlias.setBalance(0.00);
        System.out.println("\nAfter zeroing the balance through an alias:");
        carl.printAccount();
        System.out.println("carl == carlAlias ? " + (carl == carlAlias));
    }
}

Account.main(null);
```
