### My Answer to Popcorn Hack 1

**Key formatting tools:**

| Tool | What it does |
|---|---|
| `System.out.print()` | Prints and leaves the cursor on the same line |
| `System.out.println()` | Prints and moves to a new line |
| `System.out.printf()` | Prints with placeholders, so you control decimals and width |
| `%.2f` | Placeholder for a decimal rounded to 2 places |
| `%s` / `%d` | Placeholders for a String and a whole number |
| `\n` | Newline escape sequence |
| `\t` | Tab escape sequence, good for lining up columns |
| `\"` | A literal double quote inside a String |
| `\\` | A literal backslash |
| `+` | Concatenates Strings and converts numbers to text |

One line using eight of them at once:

```python
// CODE_RUNNER: Popcorn Hack 1: one line using printf, %s, %.2f, \t, \n, escaped quotes and a backslash.

public class FormattingDemo {
    public static void main(String[] args) {
        System.out.printf("Name:\t%s\nScore:\t%.2f\nQuote:\t\"nice \\ work\"\n",
                "Dhyan", 92.456);
    }
}

FormattingDemo.main(null);
```

### My Answer to Popcorn Hack 2

**Tools used in `Menu.java`:**

- `System.out.println()` for each menu line, so every option gets its own row.
- `System.out.print("Choose an option: ")` instead of `println`, so the cursor stays put and the
  user types on the same line as the prompt.
- String concatenation with `+` to drop `choice` and `optionCount` into a sentence, with Java
  converting the numbers to text automatically.
- `====` around the title as a plain text divider.

**Why it is cleaner:** the menu is scannable because each option is one line, the prompt behaves
like a real terminal prompt, and the option count comes from a variable, so the message can never
disagree with the number of options actually listed.

## Homework Hack 2
- Expand `Menu.java` into a **calculator menu**:
  1. Print a menu with options: `Add`, `Subtract`, `Multiply`, `Divide`.
  2. Ask the user for two numbers.
  3. Use arithmetic expressions to compute and display the result.
  4. Example run:
~~~
==== Calculator Menu ====
1. Add
2. Subtract
3. Multiply
4. Divide
Choose an option: 1
Enter first number: 10
Enter second number: 5
Result: 15
~~~

```python
// CODE_RUNNER: Homework Hack 2: calculator menu. Change the arguments to run() to try other operations.

// Homework Hack 2: calculator menu
// Scanner cannot read keyboard input inside Jupyter, so the math is split into
// calculate() and driven with fixed values here.

public class CalculatorMenu {

    public static double calculate(int choice, double a, double b) {
        if (choice == 1) return a + b;
        if (choice == 2) return a - b;
        if (choice == 3) return a * b;
        if (choice == 4) {
            if (b == 0) {
                System.out.println("Cannot divide by zero.");
                return 0;
            }
            return a / b;
        }
        System.out.println("Invalid option.");
        return 0;
    }

    public static void run(int choice, double first, double second) {
        System.out.println("==== Calculator Menu ====");
        System.out.println("1. Add");
        System.out.println("2. Subtract");
        System.out.println("3. Multiply");
        System.out.println("4. Divide");
        System.out.println("Choose an option: " + choice);
        System.out.println("Enter first number: " + first);
        System.out.println("Enter second number: " + second);
        System.out.println("Result: " + calculate(choice, first, second));
        System.out.println();
    }

    public static void main(String[] args) {
        run(1, 10, 5);
        run(2, 10, 5);
        run(3, 10, 5);
        run(4, 10, 5);
        run(4, 10, 0);
    }
}

CalculatorMenu.main(null);
```

The `Scanner` version below is the one I would hand in as a `.java` file. It needs a real
terminal for input.

```python
// Homework Hack 2: interactive version (terminal only)

import java.util.Scanner;

public class CalculatorMenuInteractive {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("==== Calculator Menu ====");
        System.out.println("1. Add");
        System.out.println("2. Subtract");
        System.out.println("3. Multiply");
        System.out.println("4. Divide");

        System.out.print("Choose an option: ");
        int choice = sc.nextInt();

        System.out.print("Enter first number: ");
        double first = sc.nextDouble();

        System.out.print("Enter second number: ");
        double second = sc.nextDouble();

        double result;
        if (choice == 1)      result = first + second;
        else if (choice == 2) result = first - second;
        else if (choice == 3) result = first * second;
        else if (choice == 4) result = second == 0 ? 0 : first / second;
        else {
            System.out.println("Invalid option.");
            sc.close();
            return;
        }

        System.out.println("Result: " + result);
        sc.close();
    }
}
```
