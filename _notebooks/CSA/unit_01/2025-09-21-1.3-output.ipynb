## Expressions in Java

Expressions are combinations of **literals**, **variables**, and **operators** that Java evaluates to produce a value.  
In Unit 1.3, we focus on **arithmetic expressions**.

### Arithmetic Operators
- `+` → addition  
- `-` → subtraction  
- `*` → multiplication  
- `/` → division  
- `%` → modulus (remainder)  

```java
System.out.println(7 + 3);   // 10
System.out.println(7 - 3);   // 4
System.out.println(7 * 3);   // 21
System.out.println(7 / 3);   // 2  (integer division)
System.out.println(7 % 3);   // 1  (remainder)
```

### Integer vs. Decimal Division

- If both operands are int, the result is truncated (no decimals).

- If at least one operand is a double, the result is a double.

```java
System.out.println(7 / 3);     // 2
System.out.println(7.0 / 3);   // 2.333333...
System.out.println(7 / 3.0);   // 2.333333...
```

### Operator Precedence

Java uses order of operations (PEMDAS):

- Parentheses ()

- Multiplication *, Division /, Modulus %

- Addition +, Subtraction -

```java
System.out.println(2 + 3 * 4);   // 14
System.out.println((2 + 3) * 4); // 20
```

### Combining Strings and Numbers

Using + with a string triggers string concatenation.

```java
System.out.println("Result: " + 5 + 3);   // Result: 53
System.out.println("Result: " + (5 + 3)); // Result: 
```

## Outputs

### Types of Output
- **System.out.print()** → prints text without newline.
- **System.out.println()** → prints text with newline.
- **System.out.printf()** → formatted printing (useful for decimals, alignment).

👉 Think of output as the program talking back to the user. Clear formatting is key!

```java
// Example: println vs print
System.out.print("Hello");
System.out.print(" World");
System.out.println("!"); // moves to new line
System.out.println("Done");
```

```java
// Example: Escape Sequences
System.out.println("She said: \"Java is fun!\"");
System.out.println("Line1\nLine2");
System.out.println("C:\\Users\\Student");
System.out.println("Column1\tColumn2");

```

**Key formatting tools:**

| Tool | What it does |
|---|---|
| `System.out.print()` | Prints text and leaves the cursor on the same line |
| `System.out.println()` | Prints text and moves to a new line |
| `System.out.printf()` | Prints with placeholders, so you control decimals and width |

| `%s` / `%d` | Placeholders for a String and a whole number |
| `\n` | Newline escape sequence |
| `\t` | Tab escape sequence, useful for lining up columns |
| `\"` | Prints a literal double quote inside a String |
| `\\` | Prints a literal backslash |
| `+` | Concatenates Strings, and converts numbers to text along the way |

The line of code below uses eight of them at once.

## Real-World Example — Menu.java
Menus use **expressions and output** to guide the user. This is a real-world example of combining literals, escape sequences, and arithmetic into meaningful display.
- **Output** gives users choices.
- **Expressions** let us count and customize menus.
- **Input (preview of Unit 1.4)** lets users interact with the program.

```java
import java.util.Scanner;

public class Menu {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("==== Main Menu ====");
        System.out.println("1. Start Game");
        System.out.println("2. Instructions");
        System.out.println("3. Exit");

        System.out.print("Choose an option: ");
        int choice = sc.nextInt();

        System.out.println("You selected option: " + choice);

        int optionCount = 3;
        System.out.println("There are " + optionCount + " total options.");
    }
}

Menu.main(new String[]{});
```

### My Answer to Popcorn Hack 2

**Tools used in `Menu.java`:**

- `System.out.println()` for each menu line, so every option lands on its own row.
- `System.out.print("Choose an option: ")` instead of `println`, so the cursor stays on the same
  line and the user types their answer right after the prompt. That single choice is what makes it
  look like a real prompt instead of a label floating above the input.
- String concatenation with `+` to drop the variables `choice` and `optionCount` into a sentence
  without any extra conversion code.
- `====` around the title as a plain text divider.

**Why it makes the output better:** the menu is scannable because each option is one line, the
prompt behaves the way a terminal prompt should, and the totals are built from a variable rather
than typed by hand, so the message can never disagree with the actual number of options.

### Why Menu.java Matters
- Uses **output** to display instructions.
- Uses **expressions** to dynamically show option count.
- Uses **escape sequences** for clean formatting.
- Real-world: menus are everywhere (ATMs, games, apps, vending machines).

```java
// CODE_RUNNER: Homework Hack 2: calculator menu. Change the arguments to run() to try other operations.

// Homework Hack 2: calculator menu
// Scanner does not read keyboard input inside Jupyter, so calculate() is split out
// and called with fixed values. The interactive version is in the next cell.

public class CalculatorMenu {

    public static double calculate(int choice, double a, double b) {
        if (choice == 1) return a + b;
        if (choice == 2) return a - b;
        if (choice == 3) return a * b;
        if (choice == 4) {
            if (b == 0) {
                System.out.println("Cannot divide by zero.");
                return 0;
            }
            return a / b;
        }
        System.out.println("Invalid option.");
        return 0;
    }

    public static void run(int choice, double first, double second) {
        System.out.println("==== Calculator Menu ====");
        System.out.println("1. Add");
        System.out.println("2. Subtract");
        System.out.println("3. Multiply");
        System.out.println("4. Divide");
        System.out.println("Choose an option: " + choice);
        System.out.println("Enter first number: " + first);
        System.out.println("Enter second number: " + second);
        System.out.println("Result: " + calculate(choice, first, second));
        System.out.println();
    }

    public static void main(String[] args) {
        run(1, 10, 5);
        run(2, 10, 5);
        run(3, 10, 5);
        run(4, 10, 5);
        run(4, 10, 0);
    }
}

CalculatorMenu.main(null);
```

The version below is what I would actually submit as a `.java` file. It uses `Scanner` for
real input, so it only runs from a terminal, not from inside a notebook.

```java
// Homework Hack 2: interactive version (run from a terminal, not Jupyter)

import java.util.Scanner;

public class CalculatorMenuInteractive {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("==== Calculator Menu ====");
        System.out.println("1. Add");
        System.out.println("2. Subtract");
        System.out.println("3. Multiply");
        System.out.println("4. Divide");

        System.out.print("Choose an option: ");
        int choice = sc.nextInt();

        System.out.print("Enter first number: ");
        double first = sc.nextDouble();

        System.out.print("Enter second number: ");
        double second = sc.nextDouble();

        double result;
        if (choice == 1)      result = first + second;
        else if (choice == 2) result = first - second;
        else if (choice == 3) result = first * second;
        else if (choice == 4) result = second == 0 ? 0 : first / second;
        else {
            System.out.println("Invalid option.");
            sc.close();
            return;
        }

        System.out.println("Result: " + result);
        sc.close();
    }
}
```
