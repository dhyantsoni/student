# Homework: Ultimate Battle

---

## Scenario

You are programming a game where 2 objects of your choosing (e.g., Robots, Dinosaurs, People) battle. Each object has health and power.

---

## Instructions

1. **Create a class** representing your object with:

   **Instance variables:**  
   - `String name`  
   - `int power`  
   - `int health`  
   - `MAKE YOUR OWN`

   **Static variables:**
   - `double fightDuration`

   **Instance methods:**  
   - `void attack()`
   - `void printStatus()`
   - `MAKE YOUR OWN`

   **Class methods:**  
   - `static int strongerFighter()`
   - `static void beginBattle()`
   - `MAKE YOUR OWN`

2. **In `main`:**  
   - Create two objects.  
   - Use instance methods to attack and print status.  
   - Use static methods to compare, print a fact, and start the battle.

```java
// CODE_RUNNER: Homework: Ultimate Battle. Change the power and health values and re-run the fight.

// Homework: Ultimate Battle -- two dinosaurs fight

public class Dinosaur {
    // instance variables: unique to each dinosaur
    String name;
    int power;
    int health;
    String species;              // my own instance variable

    // static variable: shared by every dinosaur
    static double fightDuration = 0.0;
    static int totalAttacks = 0;

    public Dinosaur(String name, int power, int health, String species) {
        this.name = name;
        this.power = power;
        this.health = health;
        this.species = species;
    }

    // ---------- instance methods ----------

    public void attack(Dinosaur target) {
        int damage = power + (int)(Math.random() * 10);
        target.health -= damage;
        if (target.health < 0) {
            target.health = 0;
        }
        totalAttacks++;
        fightDuration += 1.5;
        System.out.println(name + " hits " + target.name + " for " + damage + " damage!");
    }

    public void printStatus() {
        System.out.println(name + " the " + species + " -- health: " + health + ", power: " + power);
    }

    public void roar() {                     // my own instance method
        System.out.println(name + " lets out a roar that shakes the ground!");
    }

    public boolean isAlive() {
        return health > 0;
    }

    // ---------- class methods ----------

    public static int strongerFighter(Dinosaur a, Dinosaur b) {
        if (a.power > b.power) return 1;
        if (b.power > a.power) return 2;
        return 0;
    }

    public static void beginBattle(Dinosaur a, Dinosaur b) {
        fightDuration = 0.0;
        totalAttacks = 0;
        System.out.println("=== BATTLE START: " + a.name + " vs " + b.name + " ===");
    }

    public static void printBattleFact() {       // my own class method
        System.out.println("Fact: T. rex could bite with about 12,800 pounds of force.");
    }

    public static void printSummary() {
        System.out.println("Battle lasted " + fightDuration + " seconds over "
                + totalAttacks + " attacks.");
    }

    // ---------- main ----------

    public static void main(String[] args) {
        Dinosaur rex = new Dinosaur("Rex", 18, 120, "Tyrannosaurus");
        Dinosaur spike = new Dinosaur("Spike", 15, 140, "Triceratops");

        Dinosaur.printBattleFact();
        System.out.println();

        Dinosaur.beginBattle(rex, spike);

        int stronger = Dinosaur.strongerFighter(rex, spike);
        if (stronger == 1) {
            System.out.println(rex.name + " has the higher power stat.");
        } else if (stronger == 2) {
            System.out.println(spike.name + " has the higher power stat.");
        } else {
            System.out.println("Both fighters have equal power.");
        }
        System.out.println();

        rex.printStatus();
        spike.printStatus();
        System.out.println();

        rex.roar();
        int round = 1;
        while (rex.isAlive() && spike.isAlive() && round <= 6) {
            System.out.println("-- Round " + round + " --");
            rex.attack(spike);
            if (spike.isAlive()) {
                spike.attack(rex);
            }
            round++;
        }
        System.out.println();

        rex.printStatus();
        spike.printStatus();
        System.out.println();

        if (rex.isAlive() && !spike.isAlive()) {
            System.out.println(rex.name + " wins!");
        } else if (spike.isAlive() && !rex.isAlive()) {
            System.out.println(spike.name + " wins!");
        } else {
            System.out.println("Both fighters are still standing. Call it a draw.");
        }

        Dinosaur.printSummary();
    }
}

Dinosaur.main(null);
```
