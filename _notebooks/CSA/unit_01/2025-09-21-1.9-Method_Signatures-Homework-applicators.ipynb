### Popcorn Hack: Write and test max overloads
Instructions: Write two overloaded methods named max that each return the larger of two numbers.

One should accept two int values.

The other should accept two double values.

Then, test both versions in the main method.

```python
// CODE_RUNNER: Popcorn Hack: overloaded max. Which overload does max(2, 2.0) pick?

public class PopcornMax {

    public static int max(int a, int b) {
        return a >= b ? a : b;
    }

    public static double max(double a, double b) {
        return a >= b ? a : b;
    }

    public static void main(String[] args) {
        System.out.println(max(3, 9));      // expected: 9
        System.out.println(max(-2, -7));    // expected: -2
        System.out.println(max(3.5, 2.9));  // expected: 3.5
        System.out.println(max(2, 2.0));    // should call double version -> 2.0
    }
}

PopcornMax.main(null);
```

- A) `int`, `double`
- B) `double`, `double`
- C) `int`, `int`

4) Which call is ambiguous with only these two methods?
~~~java
void h(long x) {}
void h(float x) {}
~~~
- A) `h(5)`
- B) `h(5.0)`
- C) `h(5f)`
- D) None are ambiguous

5) Which statement is TRUE?
- A) Parameter names affect the method signature.
- B) Return type affects the method signature.
- C) Access modifiers affect the method signature.
- D) Overloading requires different parameter lists.


### My MCQ Answers

**1) Which pair is a valid overload? -> C**
`int f(int a)` and `double f(double a)` have different parameter types, so their signatures are
`f(int)` and `f(double)`. A is wrong because parameter *names* are not part of a signature, so both
are `f(int)`. B is wrong because you cannot overload on return type alone. D is the same method
twice.

**2) The AP CSA notion of a method signature -> B**
`max(int, int)`, that is, name plus parameter types in order. Return type, access modifiers,
parameter names, and `throws` clauses are all excluded.

**3) `p(5)` then `p(5.0)` -> A**
Prints `int`, then `double`. `5` is an `int` literal so it matches `p(int)` exactly, and `5.0` is a
`double` literal so it matches `p(double)` exactly. Neither call needs a conversion.

**4) Which call is ambiguous? -> D, none are ambiguous**
This one is a trap. `h(5)` *looks* ambiguous because an `int` can widen to both `long` and `float`,
but when several overloads apply Java picks the **most specific** one, and a `long` can itself widen
to a `float`, which makes `h(long)` more specific. So `h(5)` resolves to `h(long)`. `h(5f)` matches
`h(float)` exactly. `h(5.0)` is a `double`, and narrowing a `double` down to `long` or `float`
never happens implicitly, so that call fails to compile, but it fails for "no applicable method",
not for ambiguity.

**5) Which statement is TRUE? -> D**
Overloading requires different parameter lists. A, B, and C all name things that are deliberately
*excluded* from the signature.

### Short answer

### My Short Answers

**Why `int sum(int a, int b)` and `double sum(int a, int b)` cannot both exist**

In AP CSA, a signature is the method name plus its parameter types in order. Both of these have the
signature `sum(int, int)`, and the return type is not part of it, so as far as the compiler is
concerned they are the same method declared twice. The practical reason the rule exists is that the
call site would be undecidable: in `sum(3, 4);` written as a plain statement, the return value is
discarded, so there is nothing left to tell Java which version you meant. To overload, you have to
change the parameter list, not the return type.

**Parameters vs arguments, in one sentence**

Parameters are the typed placeholders named in the method declaration, while arguments are the
actual values handed over at the call site, so in `max(int a, int b)` called as `max(3, 9)`, `a` and
`b` are the parameters and `3` and `9` are the arguments.

```python
// CODE_RUNNER: Coding Task 1: three overloads of abs.

// Coding Task 1: three overloads of abs

public class MyAbs {
    public static int abs(int x)       { return x < 0 ? -x : x; }
    public static double abs(double x) { return x < 0 ? -x : x; }
    public static long abs(long x)     { return x < 0 ? -x : x; }

    public static void main(String[] args) {
        System.out.println(abs(-5));      // 5, calls abs(int)
        System.out.println(abs(-5.5));    // 5.5, calls abs(double)
        System.out.println(abs(-5L));     // 5, calls abs(long)
    }
}

MyAbs.main(null);
```

```python
// CODE_RUNNER: Coding Task 2: two overloads of concat.

// Coding Task 2: two overloads of concat

public class MyConcat {

    public static String concat(String a, String b) {
        return a + b;
    }

    public static String concat(String a, int n) {
        String result = "";
        for (int i = 0; i < n; i++) {
            result += a;
        }
        return result;
    }

    public static void main(String[] args) {
        System.out.println(concat("Hello, ", "world"));  // Hello, world
        System.out.println(concat("ab", 3));             // ababab
        System.out.println(concat("ab", 0));             // empty string
    }
}

MyConcat.main(null);
```

- `show(7)` prints `int`. `7` is an `int` literal and `show(int)` is an exact match, so no widening

- `show(7L)` prints `long`. The `L` suffix makes it a `long` literal, matching `show(long)` exactly.
- `show(7.0)` prints `double`. A decimal literal with no suffix is a `double`, matching

Every call has an exact match here, so overload resolution never has to fall back on widening. If
`show(long)` were deleted, `show(7L)` would widen to `show(double)` instead.

### FRQ-style
- FRQ 1. `indexOf(char target, String s)`: return the index of first `target` in `s` or `-1` if not found. Overload with `indexOf(String target, String s)` for first substring occurrence (without using library indexOf). State assumptions and constraints.
- FRQ 2. Overload `clamp`:
  - `int clamp(int value, int low, int high)` returns `value` confined to `[low, high]`.
  - `double clamp(double value, double low, double high)` similarly for doubles.
  Handle the case `low > high` by swapping.


```python
// CODE_RUNNER: FRQ 1: indexOf written without the library method. Try your own search terms.

// FRQ 1: indexOf overloads, written without calling String.indexOf
//
// Assumptions: s is not null. An empty target String matches at index 0.
// Both methods return -1 when the target is not present.

public class MyIndexOf {

    public static int indexOf(char target, String s) {
        for (int i = 0; i < s.length(); i++) {
            if (s.charAt(i) == target) {
                return i;
            }
        }
        return -1;
    }

    public static int indexOf(String target, String s) {
        if (target.length() > s.length()) {
            return -1;
        }
        for (int i = 0; i <= s.length() - target.length(); i++) {
            if (s.substring(i, i + target.length()).equals(target)) {
                return i;
            }
        }
        return -1;
    }

    public static void main(String[] args) {
        System.out.println(indexOf('r', "iterate"));       // 3
        System.out.println(indexOf('z', "iterate"));       // -1
        System.out.println(indexOf("rate", "iterate"));    // 3
        System.out.println(indexOf("era", "iterate"));     // 2
        System.out.println(indexOf("iterated", "iterate"));// -1
        System.out.println(indexOf("", "iterate"));        // 0
    }
}

MyIndexOf.main(null);
```

```python
// CODE_RUNNER: FRQ 2: clamp with bound swapping. Try passing low greater than high.

// FRQ 2: clamp overloads, swapping when low > high

public class Clamp {

    public static int clamp(int value, int low, int high) {
        if (low > high) {
            int temp = low;
            low = high;
            high = temp;
        }
        if (value < low)  return low;
        if (value > high) return high;
        return value;
    }

    public static double clamp(double value, double low, double high) {
        if (low > high) {
            double temp = low;
            low = high;
            high = temp;
        }
        if (value < low)  return low;
        if (value > high) return high;
        return value;
    }

    public static void main(String[] args) {
        System.out.println(clamp(15, 0, 10));        // 10
        System.out.println(clamp(-4, 0, 10));        // 0
        System.out.println(clamp(7, 0, 10));         // 7
        System.out.println(clamp(7, 10, 0));         // 7, bounds swapped
        System.out.println(clamp(3.7, 0.0, 1.0));    // 1.0
        System.out.println(clamp(0.5, 1.0, 0.0));    // 0.5, bounds swapped
    }
}

Clamp.main(null);
```
