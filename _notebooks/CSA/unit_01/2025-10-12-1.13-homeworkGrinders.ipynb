```python
// CODE_RUNNER: Homework Hack #1: create two Car objects and print them.

// Homework Hack #1: Object Creation Practice

public class ObjectCreation {
    public static void main(String[] args) {
        Car car1 = new Car("Tesla", 2024);
        Car car2 = new Car("Subaru", 2019);

        System.out.println(car1);
        System.out.println(car2);
    }
}

class Car {
    String brand;
    int year;

    public Car(String brand, int year) {
        this.brand = brand;
        this.year = year;
    }

    // toString() is inherited from Object; overriding it means println(car1)
    // prints this instead of something like Car@1b6d3586
    public String toString() {
        return year + " " + brand;
    }
}

ObjectCreation.main(null);
```

## Homework Hack #2 — Heap vs Stack Storage Demo

**Goal** - Understand where data is stored in memory — stack vs heap.
<br>
Instructions:
- Create a Book class with one variable: String title.
- In main(), create:
- A primitive variable (e.g. int pages = 300;)
- A Book object (e.g. Book b1 = new Book("Java Basics");)
- Copy both into new variables (int pagesCopy = pages;, Book b2 = b1;)
<br>
Change the original values and print everything — watch what changes.

```python
// CODE_RUNNER: Homework Hack #2: stack vs heap. Watch which copy changes and which does not.

// Homework Hack #2: Heap vs Stack Storage Demo

public class HeapVsStack {
    public static void main(String[] args) {
        int pages = 300;
        int pagesCopy = pages;          // copies the VALUE

        Book b1 = new Book("Java Basics");
        Book b2 = b1;                   // copies the ADDRESS, not the book

        System.out.println("Before any changes:");
        System.out.println("  pages = " + pages + ", pagesCopy = " + pagesCopy);
        System.out.println("  b1 = " + b1 + ", b2 = " + b2);

        pages = 999;
        b1.title = "Advanced Java";

        System.out.println();
        System.out.println("After changing pages and b1.title:");
        System.out.println("  pages = " + pages + ", pagesCopy = " + pagesCopy);
        System.out.println("  b1 = " + b1 + ", b2 = " + b2);

        System.out.println();
        System.out.println("pagesCopy did NOT change, because primitives live on the stack");
        System.out.println("and copying one duplicates the value.");
        System.out.println("b2 DID change, because b1 and b2 hold the same heap address,");
        System.out.println("so there was only ever one Book to edit.");
        System.out.println("b1 == b2 ? " + (b1 == b2));
    }
}

class Book {
    String title;

    public Book(String title) {
        this.title = title;
    }

    public String toString() {
        return title;
    }
}

HeapVsStack.main(null);
```
