## 📝 Homework Hack: Phone Class

Create a `Phone` class that demonstrates **all concepts** from this lesson:
- Libraries/Classes
- Packages (use ArrayList from java.util)
- Attributes
- Behaviors

### Requirements:

**Attributes (4):**
- `brand` (String)
- `model` (String)
- `batteryLevel` (int) - starts at 100
- `contacts` (ArrayList<String>)

**Behaviors (5 methods):**
1. Constructor - sets brand and model, initializes empty contacts list
2. `displayInfo()` - prints brand, model, and battery level
3. `addContact(String name)` - adds name to contacts list
4. `showContacts()` - prints all contacts
5. `usePhone(int minutes)` - decreases battery by minutes used

**Testing:**
- Create 2 Phone objects
- Add 3 contacts to each phone
- Use phone for some minutes
- Display all information

### Grading:
- Correct attributes (4 points)
- Correct methods (5 points)
- Proper use of ArrayList from java.util (3 points)
- Complete testing (3 points)
- **Total: 15 points**

```java
// CODE_RUNNER: Homework Hack: the Phone class. Add another contact or change the battery drain.

// Homework Hack: the Phone class

import java.util.ArrayList;

class Phone {
    private String brand;
    private String model;
    private int batteryLevel;
    private ArrayList<String> contacts;

    public Phone(String brand, String model) {
        this.brand = brand;
        this.model = model;
        this.batteryLevel = 100;
        this.contacts = new ArrayList<>();
    }

    public void displayInfo() {
        System.out.println(brand + " " + model + " -- battery: " + batteryLevel + "%");
    }

    public void addContact(String name) {
        contacts.add(name);
        System.out.println("Added " + name + " to " + model + "'s contacts");
    }

    public void showContacts() {
        System.out.println(model + " contacts (" + contacts.size() + "):");
        for (int i = 0; i < contacts.size(); i++) {
            System.out.println("  " + (i + 1) + ". " + contacts.get(i));
        }
    }

    public void usePhone(int minutes) {
        batteryLevel -= minutes;
        if (batteryLevel < 0) {
            batteryLevel = 0;
        }
        System.out.println("Used " + model + " for " + minutes + " minutes. Battery now "
                + batteryLevel + "%");
    }
}

public class PhoneTest {
    public static void main(String[] args) {
        Phone phone1 = new Phone("Apple", "iPhone 15");
        Phone phone2 = new Phone("Google", "Pixel 9");

        phone1.addContact("Mom");
        phone1.addContact("Dad");
        phone1.addContact("Coach Ramirez");

        phone2.addContact("Priya");
        phone2.addContact("Marcus");
        phone2.addContact("Study Group");

        System.out.println();
        phone1.usePhone(35);
        phone2.usePhone(120);

        System.out.println();
        phone1.displayInfo();
        phone1.showContacts();

        System.out.println();
        phone2.displayInfo();
        phone2.showContacts();
    }
}

PhoneTest.main(null);
```
