# Scenario

Its your first day of class when Mr. Mortenson asks you to identify all the students and list some characteristics about them. Create a class to use as a template to create student objects and create 3 imaginary students.
---

# Instructions
1. **Create a class**
    -    **Instance variables:**  
    - `String name`  
    - `int grade`  
    - `int pets`  
    - `int siblings`
    - `MAKE YOUR OWN`

2. **In `main`:**  
    - Create three student objects.  


```python
// CODE_RUNNER: Homework Scenario 1: a Student class and three student objects.

// Scenario 1: a Student class used to create three student objects

public class Student {
    String name;
    int grade;
    int pets;
    int siblings;
    String favoriteSubject;      // my own instance variable

    public Student(String name, int grade, int pets, int siblings, String favoriteSubject) {
        this.name = name;
        this.grade = grade;
        this.pets = pets;
        this.siblings = siblings;
        this.favoriteSubject = favoriteSubject;
    }

    public void printInfo() {
        System.out.println(name + " -- grade " + grade + ", " + pets + " pet(s), "
                + siblings + " sibling(s), favorite subject: " + favoriteSubject);
    }

    public static void main(String[] args) {
        Student maya = new Student("Maya Chen", 11, 2, 1, "Computer Science");
        Student jordan = new Student("Jordan Reyes", 11, 0, 3, "History");
        Student amara = new Student("Amara Okafor", 12, 1, 0, "Biology");

        maya.printInfo();
        jordan.printInfo();
        amara.printInfo();
    }
}

Student.main(null);
```

# New Scenario

You find out that one of the students that you created an object for goes by a nickname. Using your knowledge of reference variables,  create another object that references back to the same student.
---

# Instructions
1. **Use your previous code**


2. **In `main`:**  
    - Create a reference variable with a different name that points to the same student. 
    - Output the instance attributes of this student


```python
// CODE_RUNNER: Homework Scenario 2: a nickname is a second reference to the same object.

// Scenario 2: a nickname is just another reference to the SAME object

class Student {
    String name;
    int grade;
    int pets;
    int siblings;
    String favoriteSubject;

    public Student(String name, int grade, int pets, int siblings, String favoriteSubject) {
        this.name = name;
        this.grade = grade;
        this.pets = pets;
        this.siblings = siblings;
        this.favoriteSubject = favoriteSubject;
    }

    public void printInfo() {
        System.out.println(name + " -- grade " + grade + ", " + pets + " pet(s), "
                + siblings + " sibling(s), favorite subject: " + favoriteSubject);
    }
}

public class NicknameDemo {
    public static void main(String[] args) {
        Student maya = new Student("Maya Chen", 11, 2, 1, "Computer Science");

        // no 'new' here, so no second student is created
        Student mimi = maya;

        System.out.println("Through the nickname reference:");
        mimi.printInfo();

        mimi.favoriteSubject = "Robotics";

        System.out.println();
        System.out.println("After changing it through 'mimi':");
        System.out.print("maya -> ");
        maya.printInfo();
        System.out.print("mimi -> ");
        mimi.printInfo();

        System.out.println();
        System.out.println("maya == mimi ? " + (maya == mimi));
    }
}

NicknameDemo.main(null);
```
