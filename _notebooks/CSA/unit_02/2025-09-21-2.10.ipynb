<h3> Finding a substring within a string </h3>

We can use the "window" method:

A "window" is created the same length as the substring. We then iterate through the main string in sections and compare to the substring

For example:

<h3> I T E R A T E </h3>with substring "ERA"

<br>
<br>
<br>

````python
public class StringFinder {
    public static void main(String[] args) {
        String word = "iterate";
        String sub = "xxx";
        boolean found = false; // will be set to true once substring is found

        for (int i = 0; i < word.length(); i++) { //iterating forwards: starting at first index (0) and going to the length of the word.. let's try word.length
            String portion = word.substring(i, i + sub.length());
            if (sub.length() > portion.length()) break;
            if (portion.equals(sub)) { // make sure you use .equals!!
                found = true;
                break;
            }
        }

        if (found)
            System.out.println("substring is found within string!");
        else
            System.out.println("substring is NOT within string");
    }

}


StringFinder.main(null);
````

<h4> POPCORN HACK: Run the code.. what happened? How can we fix it?</h4>

Tell us below!

### What happened

It **crashed** with a `StringIndexOutOfBoundsException` instead of printing "substring is NOT
within string".

The loop runs `i` from `0` to `word.length() - 1` (0 to 6). But the very first thing in the body is

~~~java
String portion = word.substring(i, i + sub.length());
~~~

Once `i` reaches 5, that asks for `substring(5, 8)` on a 7-character word - past the end of the
string. The `if (sub.length() > portion.length()) break;` guard was supposed to stop that, but it
runs *after* the `substring` call, so it never gets the chance.

### How to fix it

Stop the loop before the window runs off the end. The last valid starting index is
`word.length() - sub.length()`, so the condition should be:

~~~java
for (int i = 0; i <= word.length() - sub.length(); i++)
~~~

<br>

<h4> Another issue:</h4>

<h3> I T E R A T E </h3>
What if our substring was the word "RATE"? Note that RATE is at the end of the whole string

### Answer

That is exactly why the bound has to be `i <= word.length() - sub.length()` and **not**
`i < word.length() - sub.length()`. For `"iterate"` (length 7) and `"rate"` (length 4), "rate"
starts at index 3, and `7 - 4 = 3`. Using `<` would stop at `i = 2` and miss it. The `<=` keeps the
final window in play.

(One more gotcha: `.equals()` is case sensitive, so `"RATE"` will not match `"rate"`. Either
compare with `.equalsIgnoreCase()` or lowercase both strings first.)

HINT: Use the boolean expressions we have learned in previous lessons. Which would you use when comparing your "window" with multiple options of substrings?

### My Answer

A vowel is a window of length 1, so I iterate over every index, pull out a one-character
substring, and use a **compound OR (`||`)** to compare it against all five vowels at once. `||`
short-circuits, so as soon as one comparison is true Java stops checking the rest.

My solution is two cells down.

````python
// CODE_RUNNER: Popcorn hack: the loop bound is fixed. Try sub = "xxx" and sub = "era".

// POPCORN HACK -- the fixed version of StringFinder

public class StringFinderFixed {
    public static void main(String[] args) {
        String word = "iterate";
        String sub = "rate";       // try "xxx" and "era" too
        boolean found = false;

        // the last legal starting index is word.length() - sub.length()
        for (int i = 0; i <= word.length() - sub.length(); i++) {
            String portion = word.substring(i, i + sub.length());
            if (portion.equals(sub)) {
                found = true;
                System.out.println("Found \"" + sub + "\" starting at index " + i);
                break;
            }
        }

        if (found)
            System.out.println("substring is found within string!");
        else
            System.out.println("substring is NOT within string");
    }
}

StringFinderFixed.main(null);
````

### HACKS: count the vowels using iteration

````python
// CODE_RUNNER: HACKS: count the vowels in a string. Try your own words.

// HACKS -- count the number of vowels in a string using iteration

public class VowelCounter {

    public static int countVowels(String s) {
        int count = 0;
        String lower = s.toLowerCase();

        for (int i = 0; i < lower.length(); i++) {
            String window = lower.substring(i, i + 1);

            if (window.equals("a") || window.equals("e") || window.equals("i")
                    || window.equals("o") || window.equals("u")) {
                count++;
            }
        }
        return count;
    }

    public static void main(String[] args) {
        System.out.println("iterate has " + countVowels("iterate") + " vowels");
        System.out.println("Computer Science has " + countVowels("Computer Science") + " vowels");
        System.out.println("rhythm has " + countVowels("rhythm") + " vowels");
        System.out.println("AEIOU has " + countVowels("AEIOU") + " vowels");
    }
}

VowelCounter.main(null);
````
