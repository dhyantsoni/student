# Informal Code Analysis

<b>Learning objective</b>: Compute statement execution counts & informal run-time comparison of iterative statements

<b>Essential Knowledge</b>: A statement execution count indicates the number of times a statement is executed by the program

<h3> What IS informal code analysis? </h3>

Answer:

**Informal code analysis is counting how many times each statement in a program runs, so you can
compare two solutions without actually timing them.**

Instead of measuring wall-clock seconds (which changes with your computer, the JVM, what else is
running, etc.), you read the loop headers and work out the *statement execution count*. Whichever
version executes fewer statements for the same input is the faster algorithm.

Why the College Board cares:
- It is machine independent - the count is the same on any computer.
- It scales: for a single loop the count is roughly `n`; for a nested loop it is roughly `n * m`.
- It gives you a way to justify "this rewrite is faster" with a number instead of a guess.

```python
// CODE EXAMPLE #1 (for loop)
public class InformalCodeAnalysis {
    public static void main(String[] args) {
        int count = 0;
        for (int k = 0; k < 30; k++)
        {
            if (k % 3 == 0) // statement 1
            {
                count++; // statement 2
            }
        }
    }
}
```

<b>How many times will statement 1 execute? </b>

Answer: **30 times.**

Statement 1 is the `if (k % 3 == 0)` condition itself, and it lives directly in the loop body.
The loop header `for (int k = 0; k < 30; k++)` runs the body once for every `k` from 0 to 29, so
the condition is evaluated on all 30 passes.

<b>How many times will statement 2 execute?</b>

Answer: **10 times.**

Statement 2 (`count++`) only runs on the passes where `k % 3 == 0` is true. Between 0 and 29 the
multiples of 3 are 0, 3, 6, 9, 12, 15, 18, 21, 24, 27 - that is 10 values, so `count` ends at 10.

```python
// CODE_RUNNER: CODE EXAMPLE #2: run it to confirm statement 3 executes 9 times.

// CODE EXAMPLE #2 (for loop)
public class InformalCodeAnalysis2 {
    public static void main(String[] args) {
        int count = 0;
        for (int k = 4; k < 30; k += 3)
        {
            count++; // statement 3
        }
        System.out.println("statement 3 ran " + count + " times");
    }
}
InformalCodeAnalysis2.main(null);
```

<b>How many times will statement 3 execute?</b>

Answer: **9 times.**

In CODE EXAMPLE #2 above, `k` starts at 4 and increases by 3 as long as `k < 30`, so it takes the
values 4, 7, 10, 13, 16, 19, 22, 25, 28 - 9 values in total. (28 + 3 = 31 fails the test, so the
loop stops there.)

Shortcut for any counted for loop: `ceil((stop - start) / step)` = `ceil((30 - 4) / 3)` =
`ceil(8.67)` = **9**.

```python
// CODE_RUNNER: Rewrite for a faster run time based on statement execution counts.

// Stepping by 5 removes the if-check: 27 executions instead of
// 135 loop passes plus 135 condition checks.

public class FasterLoop {
    public static void main(String[] args) {
        for (int k = 0; k < 135; k += 5)
        {
            System.out.println(k);
        }
    }
}

FasterLoop.main(null);
```

```python
// CODE EXAMPLE #3 (while loop)

int num = (int)(Math.random() * 10);
while (num % 2 != 0)
{
    num = (int)(Math.random() * 10); // statement 4
}
```

<b>What is the min/max number of times statement 4 will execute?</b>

Answer:

- **Minimum: 0 times.** `(int)(Math.random() * 10)` produces an int from 0 to 9. If the first draw
  is already even (0, 2, 4, 6, 8), then `num % 2 != 0` is false right away and the body never runs.
- **Maximum: unbounded - there is no maximum.** Each pass has a 5-in-10 chance of drawing another
  odd number, so the loop could in principle repeat forever. You cannot name a number it is
  guaranteed to stay under; it just gets less and less likely.

That gap between "0" and "no upper bound" is the reason this has to be a `while` loop instead of a
`for` loop - we do not know the iteration count ahead of time.

```python
// CODE EXAMPLE #4 (nested loop)
// Statement #5 executes 3 * 4 = 12 times.

for (int outer = 0; outer < 3; outer++)
{
    for (int inner = 0; inner < 4; inner++)
    {
        // statement #5
    }
}
```
